```python
import numpy as np
from sklearn.cluster import KMeans

X = np.array([
    [1, 1],
    [1.5, 2],
    [3, 4],
    [6, 7],
    [3.5, 5],
    [4.5, 5]
])

kmeans = KMeans(n_clusters=3, random_state=5, n_init=10)

kmeans.fit(X)

print("Cluster labels:")
print(kmeans.labels_)

print("Centroids:")
print(kmeans.cluster_centers_)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Student marks: [Math, Science]
X = [
    [20, 25],
    [22, 27],
    [24, 26],
    [80, 85],
    [82, 88],
    [85, 90]
]

# Create K-Means model
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

# Train the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

print("Cluster Labels:", labels)

print("Cluster Centers:")
print(kmeans.cluster_centers_)

# Plot the clusters
plt.scatter(
    [x[0] for x in X],
    [x[1] for x in X],
    c=labels
)

# Plot cluster centers
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200
)

plt.title("Student Marks Clustering")
plt.xlabel("Math Marks")
plt.ylabel("Science Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Student data: [Study Hours, Marks]
X = [
    [1, 40],
    [1.2, 42],
    [1.4, 41],

    [3, 55],
    [3.2, 57],
    [3.4, 56],

    [5, 70],
    [5.2, 72],
    [5.4, 71],

    [7, 85],
    [7.2, 87],
    [7.4, 86]
]

# Create K-Means model
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

# Train the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

print("Cluster Labels:", labels)

# Get cluster centers
print("Cluster Centers:")
print(kmeans.cluster_centers_)

# Plot the clusters
plt.scatter(
    [x[0] for x in X],
    [x[1] for x in X],
    c=labels
)

# Plot cluster centers
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200
)

plt.title("K-Means Clustering - 4 Clusters")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# House data: [House Size, Price]
X = [
    [500, 25],
    [520, 27],
    [540, 26],

    [800, 40],
    [820, 42],
    [840, 41],

    [1200, 65],
    [1220, 67],
    [1240, 66],

    [1600, 90],
    [1620, 92],
    [1640, 91]
]

# Create K-Means model
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

# Train the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

print("Cluster Labels:", labels)

# Get cluster centers
print("Cluster Centers:")
print(kmeans.cluster_centers_)

# Plot the clusters
plt.scatter(
    [x[0] for x in X],
    [x[1] for x in X],
    c=labels
)

# Plot cluster centers
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200
)

plt.title("K-Means House Clustering - 4 Clusters")
plt.xlabel("House Size (sq ft)")
plt.ylabel("Price (Lakhs)")

plt.show()
```
