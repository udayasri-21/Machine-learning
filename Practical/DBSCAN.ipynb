```python
from sklearn.cluster import DBSCAN

# Data
X = [[1], [2], [3], [10], [11], [12], [30]]

# Create DBSCAN model
model = DBSCAN(eps=2, min_samples=2)

# Fit the model
model.fit(X)

# Print labels
print("Labels:", model.labels_)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# Student data: [Attendance, Marks]
X = [
    [75, 65],
    [78, 68],
    [80, 70],
    [82, 72],

    [90, 85],
    [92, 88],
    [94, 90],
    [95, 92],

    [30, 20]       # Noise student
]

# Create DBSCAN model
dbscan = DBSCAN(eps=5, min_samples=3)

# Train the model
dbscan.fit(X)

# Get cluster labels
labels = dbscan.labels_

print("Cluster Labels:", labels)

# Plot the clusters
plt.scatter(
    [x[0] for x in X],
    [x[1] for x in X],
    c=labels
)

plt.title("Student Attendance and Marks Clustering")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# Employee data: [Experience, Salary]
X = [
    # Cluster 1
    [1.0, 25],
    [1.2, 26],
    [1.1, 25],

    # Cluster 2
    [3.0, 40],
    [3.2, 41],
    [3.1, 40],

    # Cluster 3
    [5.0, 55],
    [5.2, 56],
    [5.1, 55],

    # Cluster 4
    [7.0, 70],
    [7.2, 71],
    [7.1, 70]
]

# Create DBSCAN model
dbscan = DBSCAN(eps=0.5, min_samples=3)

# Train the model
dbscan.fit(X)

# Get cluster labels
labels = dbscan.labels_

print("Cluster Labels:", labels)

# Plot the clusters
plt.scatter(
    [x[0] for x in X],
    [x[1] for x in X],
    c=labels
)

plt.title("DBSCAN - Employee Clustering")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (Thousands)")

plt.show()
```
